```python
# dashboard 1   - 2 grafy počet případů celkem a počet úmrtí celkem
import pandas as pd
import plotly.express as px
from dash import Dash, dcc, html, Input, Output

# Načtení dat
df = pd.read_csv(r"D:\CodersLab\project_1_python.csv")
df["date"] = pd.to_datetime(df["date"], errors="coerce")

# Dash aplikace
app = Dash(__name__)

# Layout
app.layout = html.Div([
    html.H1("COVID-19 Dashboard"),

    html.Label("Select a country:"),
    dcc.Dropdown(
        id="country-filter",
        options=[{"label": c, "value": c} for c in df["location"].unique()],
        value="Czechia"  # default
    ),

    html.Div([
        dcc.Graph(id="cases-graph"),
        dcc.Graph(id="deaths-graph")
    ], style={"display": "flex"})
])

# Callback
@app.callback(
    [Output("cases-graph", "figure"),
     Output("deaths-graph", "figure")],
    [Input("country-filter", "value")]
)
def update_graphs(country):
    dff = df[df["location"] == country]

    fig_cases = px.line(
        dff, x="date", y="total_cases",
        title=f"Cumulative number of cases in {country}"
    )

    fig_deaths = px.line(
        dff, x="date", y="total_deaths",
        title=f"Cumulative number of deaths in {country}"
    )

    return fig_cases, fig_deaths

# Spuštění aplikace (ve VS Code)
if __name__ == "__main__":
    app.run(debug=True)
```
