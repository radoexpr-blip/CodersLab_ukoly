# Introduction

The purpose of this task is to extract and save information about hockey teams into JSON format, based on data from files located in the `/data/raw` directory, which were generated in the previous stage. The information to be scraped and saved includes:  

- Team Name (`Team Name`),  
- Year (`Year`),  
- Number of wins (`Wins`),  
- Number of losses (`Losses`),  
- Number of overtime losses (`OT Losses` - Overtime Losses),  
- Win percentage (`Win %`),  
- Number of goals scored (`Goals For (GF)`),  
- Number of goals conceded (`Goals Against (GA)`),  
- Goal differential (`+ / -`).  

Each collected record should be organized into a dictionary with the structure shown below and then added to the results list:  

```python  
{  
    'Team Name': 'Boston Bruins',  
    'Year': '1990',  
    'Wins': '44',  
    'Losses': '24',  
    'OT Losses': '',  
    'Win %': '0.55',  
    'Goals For (GF)': '299',  
    'Goals Against (GA)': '264',  
    '+ / -': '35'  
}  
```

Place each item into the results list.

The resulting data should be saved in a file named hockey_teams.json, which will be placed in the `data/interim/` folder. This file will serve as a data source for further analysis in the next part of the workshop.

> At this point, converting HTML to JSON may seem complex and unnecessary, but it aims to consolidate knowledge regarding this data structure due to its universality and prevalence not only in the world of data analysis but generally in IT as well.

## Import Required Libraries

````python
import json
import os
from bs4 import BeautifulSoup

html_path = r"C:\Users\rdpre\Workshop_-_files\data\raw\hockey_teams_page_001.html"
json_path = r"C:\Users\rdpre\Workshop_-_files\data\interim\hockey_teams.json"  
  
#  Načtení HTML souboru
with open(html_path, "r", encoding="utf-8") as f:
    html = f.read()
# „Vezmi ten text a udělej z něj strom HTML značek, se kterým můžu pracovat.“
soup = BeautifulSoup(html, "html.parser")

# NaJDI VŠEChno o týmech
teams = soup.find_all("tr", class_="team")

results = []
# smyčka
for team in teams:

    team_name = team.find("td", class_="name").text.strip()
    year = team.find("td", class_="year").text.strip()    
    wins = team.find("td", class_="wins").text.strip()    
    losses = team.find("td", class_="losses").text.strip()    
    ot_losses = team.find("td", class_="ot-losses").text.strip() 
    win_pct = team.find("td", class_="pct").text.strip()   
    goals_for = team.find("td", class_="gf").text.strip()    
    goals_against = team.find("td", class_="ga").text.strip()    
    diff_goals = team.find("td", class_="diff").text.strip()    
    
# ulož do slovníku

results.append({
    "Team Name": team_name,
    "Year": year,
    "Wins": wins,
    "Losses": losses,
    "OT Losses": ot_losses,
    "Win %": win_pct,
    "Goals for": goals_for,
    "Goals against": goals_against,
    "+ / -": diff_goals
  })

with open(json_path, "w", encoding="utf-8") as f:
    json.dump(results, f, indent=2, ensure_ascii=False)


print(results)
````

## List of HTML files

Using the `glob` module, find all `html` files in the `data/raw` folder.

````python
import json
from bs4 import BeautifulSoup

# Cesty k souborům
html_path = r"C:\Users\rdpre\Workshop_-_files\data\raw\hockey_teams_page_001.html"
json_path = r"C:\Users\rdpre\Workshop_-_files\data\interim\hockey_teams.json"

# Načtení HTML souboru
with open(html_path, "r", encoding="utf-8") as f:
    html = f.read()

# Vytvoření stromu HTML
soup = BeautifulSoup(html, "html.parser")

# Najdi všechny řádky týmů
teams = soup.find_all("tr", class_="team")
print(f"🔍 Počet nalezených týmů: {len(teams)}")

# Výsledný seznam
results = []

# Smyčka přes každý tým
for i, team in enumerate(teams):
    try:
        team_name = team.find("td", class_="name").text.strip()
        year = team.find("td", class_="year").text.strip()
        wins = team.find("td", class_="wins").text.strip()
        losses = team.find("td", class_="losses").text.strip()
        ot_losses = team.find("td", class_="ot-losses").text.strip()
        win_pct = team.find("td", class_="pct").text.strip()
        goals_for = team.find("td", class_="gf").text.strip()
        goals_against = team.find("td", class_="ga").text.strip()
        diff_goals = team.find("td", class_="diff").text.strip()

        # Přidání do výsledků
        results.append({
            "Team Name": team_name,
            "Year": year,
            "Wins": wins,
            "Losses": losses,
            "OT Losses": ot_losses,
            "Win %": win_pct,
            "Goals for": goals_for,
            "Goals against": goals_against,
            "+ / -": diff_goals
        })

        # Volitelný výpis každého týmu
        print(f"{i+1}. {team_name} ({year}) – W:{wins} L:{losses} OT:{ot_losses}")

    except AttributeError:
        print(f"⚠️ Chyba při zpracování týmu č. {i+1} – některá buňka chybí")

# Uložení do JSON
with open(json_path, "w", encoding="utf-8") as f:
    json.dump(results, f, indent=2, ensure_ascii=False)

print(f"\n✅ Hotovo. Uloženo {len(results)} týmů do: {json_path}")

````

````python
import json
from bs4 import BeautifulSoup

# Cesty k souborům
html_path = r"C:\Users\rdpre\Workshop_-_files\data\raw\hockey_teams_page_001.html"
json_path = r"C:\Users\rdpre\Workshop_-_files\data\interim\hockey_teams.json"

# Načtení HTML souboru
with open(html_path, "r", encoding="utf-8") as f:
    html = f.read()

# Vytvoření stromu HTML
soup = BeautifulSoup(html, "html.parser")

# Najdi všechny řádky týmů
teams = soup.find_all("tr", class_="team")
print(f"🔍 Počet nalezených týmů: {len(teams)}")

# Výsledný seznam
results = []

# Smyčka přes každý tým
for i, team in enumerate(teams):
    try:
        team_name = team.find("td", class_="name").text.strip()
        year = team.find("td", class_="year").text.strip()
        wins = team.find("td", class_="wins").text.strip()
        losses = team.find("td", class_="losses").text.strip()
        ot_losses = team.find("td", class_="ot-losses")
        ot_losses = ot_losses.text.strip() if ot_losses else ""  # prázdné pokud chybí
        win_pct = team.find("td", class_="pct").text.strip()
        goals_for = team.find("td", class_="gf").text.strip()
        goals_against = team.find("td", class_="ga").text.strip()
        diff_goals = team.find("td", class_="diff").text.strip()

        # Přesně požadovaná struktura
        results.append({
            "Team Name": team_name,
            "Year": year,
            "Wins": wins,
            "Losses": losses,
            "OT Losses": ot_losses,
            "Win %": win_pct,
            "Goals For (GF)": goals_for,
            "Goals Against (GA)": goals_against,
            "+ / -": diff_goals
        })

    except AttributeError:
        print(f"⚠️ Chyba při zpracování týmu č. {i+1} – některá buňka chybí")

# Uložení do JSON
with open(json_path, "w", encoding="utf-8") as f:
    json.dump(results, f, indent=2, ensure_ascii=False)

print(f"\n✅ Hotovo. Uloženo {len(results)} týmů do: {json_path}")
````
