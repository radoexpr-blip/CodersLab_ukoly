# Fetching Website Content

This section of the notebook contains code for fetching website content. To properly execute the task, consider the following steps:  
- Ensure all available data on the site has been fetched by checking if there are additional data pages.  
- Locate the data of interest on the page using `html` inspection tools.  
- Navigate between subsequent data pages using browser mechanisms or by analyzing the `url` structure.  
  
> Remember to respect the query limits specified in the `FAQ`!  
  
Save the fetched data to the folder `data/raw/hockey_teams_page_{page_number}.html`. At this stage, we are retrieving data without processing it - analysis will be performed later.  
  
To fetch the `html` content of the page, you can use `browser.page_source`. Make sure the browser tool configuration (e.g., Selenium) is ready for use.  
  
> (Optional) If there are multiple pages to fetch, use the [zfill](https://www.programiz.com/python-programming/methods/string/zfill) function to maintain order in file names by adding leading zeros to the page numbers.



```python
import requests
from bs4 import BeautifulSoup
import os # slouží k práci s OS
import json

# URL s parametrem per_page=1000
url = "http://www.scrapethissite.com/pages/forms/?page_num=1&per_page=1000"

# Výstupní složka a soubor
output_dir = r"C:\Users\rdpre\Workshop_-_files\data\raw"


os.makedirs(output_dir, exist_ok=True)
output_file = os.path.join(output_dir, "hockey_teams_page_001.html")

# Stažení HTML obsahu
response = requests.get(url)
html_content = response.text

# Uložení HTML do souboru
with open(output_file, "w", encoding="utf-8") as f:
    f.write(html_content)

print(f" HTML stránka uložena do: {output_file}")

def get_teams():
    r = requests.get("http://www.scrapethissite.com/pages/forms/?page_num=1&per_page=1000")
    soup = BeautifulSoup(r.text, "html.parser")
    table = soup.find("table", class_="table")
    return table.find_all("tr", class_="team")


teams = get_teams()
                                # <tr class="team">

                                    #<td class="name">Boston Bruins</td>

                    # Team Name 	Year 	Wins 	Losses 	OT Losses 	Win % 	Goals For (GF) 	Goals Against (GA) 	+ / - 
for team in teams:

    team_name = team.find("td", class_="name").text.strip()
    year = team.find("td", class_="year").text.strip()    
    wins = team.find("td", class_="wins").text.strip()    
    losses = team.find("td", class_="losses").text.strip()    
    ot_losses = team.find("td", class_="ot-losses").text.strip() 
    win_pct = team.find("td", class_="pct").text.strip()   
    goals_for = team.find("td", class_="gf").text.strip()    
    goals_against = team.find("td", class_="ga").text.strip()    
    diff_goals = team.find("td", class_="diff").text.strip()    
    



# print(categories_urls)
    print(f"{team_name} ({year}) | W:{wins} L:{losses} OT:{ot_losses} Win%:{win_pct} GF:{goals_for} GA:{ goals_against} +/-:{diff_goals}")

    
```
